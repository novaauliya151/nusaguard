```python
import pandas as pd

df = pd.read_csv("../data/raw/sms_spam_indo.csv")

print("Jumlah baris:", len(df))
print("\nDistribusi Kategori:")
print(df["Kategori"].value_counts())
print("\nPersentase:")
print(df["Kategori"].value_counts(normalize=True) * 100)

print("\nContoh 5 pesan per kategori:")
for kategori in df["Kategori"].unique():
    print(f"\n--- {kategori} ---")
    print(df[df["Kategori"] == kategori]["Pesan"].head(3).tolist())

print("\nPanjang teks (kata):")
print(df["Pesan"].str.split().str.len().describe())

print("\nCek data kosong/duplikat:")
print("Missing:", df.isnull().sum())
print("Duplikat:", df.duplicated().sum())
```
